```python
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
stock_data = yf.download('AAPL', start='2024-01-01')
stock_data.head(10)
```

```python
# Closing price select karo
close_price = stock_data['Close']
```

```python

# Missing values remove karo
close_price = close_price.dropna()
```

```python
close_price.head()
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose

decomposition = seasonal_decompose(
    close_price,
    model='additive',
    period=30
)
```

```python
trend = decomposition.trend
seasonality = decomposition.seasonal
residual = decomposition.resid
```

```python
plt.figure(figsize=(14, 5))
plt.plot(trend)
plt.title('AAPL Stock Price - Trend')
plt.xlabel('Date')
plt.ylabel('Price')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(14, 5))
plt.plot(seasonality)
plt.title('AAPL Stock Price - Seasonality')
plt.xlabel('Date')
plt.ylabel('Seasonal Component')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(14, 5))
plt.plot(residual)
plt.title('AAPL Stock Price - Residuals')
plt.xlabel('Date')
plt.ylabel('Residual')
plt.grid(True)
plt.show()
```
